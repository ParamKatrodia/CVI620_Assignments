## Part I - Shapes: OpenCV Logo Recreation


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

canvas = np.ones((500, 500, 3), dtype=np.uint8) * 255
center = (250, 250)
colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255)]
angles = [0, 120, 240]

for color, angle in zip(colors, angles):
    cv2.ellipse(canvas, center, (100, 160), angle, 0, 300, color, -1)

cv2.putText(canvas, 'OpenCV', (150, 470), cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 0, 0), 3, cv2.LINE_AA)

display_img = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6,6))
plt.imshow(display_img)
plt.axis('off')
plt.title('OpenCV Logo Recreation')
plt.show()
```

## Part II - Image Arithmetic (Invisible Cloak)


```python
import cv2
import numpy as np
import time

def invisible_cloak_red():
    cap = cv2.VideoCapture(0)
    time.sleep(3)

    print("Capturing background... please move out of frame.")
    for i in range(60):
        ret, background = cap.read()
        if not ret:
            continue
    background = np.flip(background, axis=1)

    print("Starting Invisible Cloak (Red) ... Press 'q' to quit.")

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame = np.flip(frame, axis=1)

        hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

        # --- Red color detection uses two ranges ---
        lower_red1 = np.array([0, 120, 70])
        upper_red1 = np.array([10, 255, 255])
        lower_red2 = np.array([170, 120, 70])
        upper_red2 = np.array([180, 255, 255])

        mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
        mask2 = cv2.inRange(hsv, lower_red2, upper_red2)
        mask = mask1 + mask2

        # Refine mask
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8), iterations=2)
        mask = cv2.morphologyEx(mask, cv2.MORPH_DILATE, np.ones((3,3), np.uint8), iterations=1)

        mask_inv = cv2.bitwise_not(mask)

        # Segment out red part and replace with background
        res1 = cv2.bitwise_and(frame, frame, mask=mask_inv)
        res2 = cv2.bitwise_and(background, background, mask=mask)
        final_output = cv2.addWeighted(res1, 1, res2, 1, 0)

        cv2.imshow("Invisible Cloak (Red)", final_output)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

# Run the function
invisible_cloak_red()
```

## Part III - A Photo Booth Application


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_image(img, title='Result'):
    plt.figure(figsize=(8,6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title(title)
    plt.show()

def adjust_brightness(img, value):
    new_img = cv2.convertScaleAbs(img, alpha=1, beta=value)
    operations.append(f"brightness {value:+}")
    history.append(new_img.copy())
    return new_img

def adjust_contrast(img, factor):
    new_img = cv2.convertScaleAbs(img, alpha=factor, beta=0)
    operations.append(f"contrast x{factor}")
    history.append(new_img.copy())
    return new_img

def convert_grayscale(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    new_img = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
    operations.append("converted to grayscale")
    history.append(new_img.copy())
    return new_img

def add_padding(img, size, border_type='constant'):
    if border_type == 'constant':
        new_img = cv2.copyMakeBorder(img, size, size, size, size,
                                     cv2.BORDER_CONSTANT, value=(0,0,0))
    elif border_type == 'reflect':
        new_img = cv2.copyMakeBorder(img, size, size, size, size, cv2.BORDER_REFLECT)
    else:
        new_img = cv2.copyMakeBorder(img, size, size, size, size, cv2.BORDER_REPLICATE)
    operations.append(f"padded {size}px {border_type}")
    history.append(new_img.copy())
    return new_img

def apply_threshold(img, mode='binary'):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    if mode == 'binary':
        _, th = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY)
    else:
        _, th = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV)
    new_img = cv2.cvtColor(th, cv2.COLOR_GRAY2BGR)
    operations.append(f"threshold {mode}")
    history.append(new_img.copy())
    return new_img

def blend_images(img1, img2_path, alpha):
    img2 = cv2.imread(img2_path)
    if img2 is None:
        print(f"Could not find {img2_path}")
        return img1
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
    new_img = cv2.addWeighted(img1, alpha, img2, 1 - alpha, 0)
    operations.append(f"blended with {img2_path}, alpha={alpha}")
    history.append(new_img.copy())
    return new_img

def undo_last():
    if len(history) > 1:
        history.pop()
        operations.pop()
        print("↩Undid last operation.")
        return history[-1]
    else:
        print("Nothing to undo.")
        return history[-1]

def view_history():
    print("\n=== History of Operations ===")
    if not operations:
        print("No operations performed yet.")
    else:
        for i, op in enumerate(operations, 1):
            print(f"{i}. {op}")
    print("=============================\n")

def photo_booth_app():
    global history, operations
    operations = []

    image = cv2.imread("S9_Lambo.png")   
    if image is None:
        print("Could not find 'S9_Lambo.png'. Please place it in this folder.")
        return
    history = [image.copy()]
    print("Image 'S9_Lambo.png' loaded successfully.\n")

    while True:
        print("""
==== Mini Photo Editor ====
1. Adjust Brightness
2. Adjust Contrast
3. Convert to Grayscale
4. Add Padding (choose border type)
5. Apply Thresholding (binary or inverse)
6. Blend with Another Image
7. Undo Last Operation
8. View History of Operations
9. Save and Exit
===========================
""")
        choice = input("Enter your choice (1-9): ")

        if choice == '1':
            value = int(input("Enter brightness value (-100 to 100): "))
            image = adjust_brightness(image, value)
            show_image(image, f"Brightness {value:+}")

        elif choice == '2':
            factor = float(input("Enter contrast factor (e.g., 0.8–1.5): "))
            image = adjust_contrast(image, factor)
            show_image(image, f"Contrast x{factor}")

        elif choice == '3':
            image = convert_grayscale(image)
            show_image(image, "Grayscale")

        elif choice == '4':
            size = int(input("Enter padding size (px): "))
            border_type = input("Border type (constant / reflect / replicate): ").strip().lower()
            image = add_padding(image, size, border_type)
            show_image(image, f"Padding {size}px {border_type}")

        elif choice == '5':
            mode = input("Threshold mode (binary / binary_inv): ").strip().lower()
            image = apply_threshold(image, mode)
            show_image(image, f"Threshold {mode}")

        elif choice == '6':
            path = "rose.png"  # Second image for blending
            alpha = float(input("Enter alpha (0–1): "))
            image = blend_images(image, path, alpha)
            show_image(image, f"Blended with {path}")

        elif choice == '7':
            image = undo_last()
            show_image(image, "After Undo")

        elif choice == '8':
            view_history()

        elif choice == '9':
            filename = input("Enter filename to save (e.g., edited.jpg): ")
            cv2.imwrite(filename, image)
            print(f"💾 Image saved as {filename}")
            view_history()
            print("Exiting Mini Photo Editor.")
            break

        else:
            print("Invalid choice. Try again.")

# Run the program
photo_booth_app()
```
